```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
```

# Dataset

```python
x,y = make_regression(n_samples= 100, n_features = 1, noise = 10)
plt.scatter(x,y)
```

```python
print(x.shape)
y = y.reshape(y.shape[0],1)
print(y.shape)
```

```python
X = np.hstack([x,np.ones(x.shape)])
X.shape
```

```python
theta = np.random.randn(2,1)
theta
```

# modele

```python
def model(X ,theta):
    return X.dot(theta)
```

```python
model(X,theta)
plt.scatter(x,y)
plt.plot(x,model(X,theta))
```

# fonction cout (mse)

```python
def cost_function(X,y,theta):
    m = len(y)
    return (1/(2*m)) * (np.sum((model(X,theta) - y) **2))
```

```python
cost_function(X,y,theta)
```

# Gradients et Descente de Gradient

```python
def grad(X,y,theta):
    m = len(y)
    return (1/m) * (X.T.dot(model(X,theta) - y))
```

```python
costs = [] #for mesure the error
def gradient_descent(X,y,theta, alpha, n_iteration):
    for i in range(0, n_iteration):
        theta = theta - alpha * grad(X,y,theta)
        costs.append(cost_function(X,y,theta))
    return theta
```

# entrainement

```python
theta_final = gradient_descent(X,y,theta, alpha = 0.2, n_iteration = 1000)
theta_final
```

```python
predictions = model(X,theta_final)
mse = cost_function(X,y,theta_final)
print(mse)
plt.scatter(x,y)
plt.plot(x,predictions)
np.mean(y)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y, model(X, theta_final))
print(r2)

def score(X,y,theta_final):
    numerator = np.sum((model(X,theta_final) - y) ** 2)
    denom = np.sum((np.mean(y) - y) **2)
    return (1 - (numerator/denom))

score(X,y,theta_final)
```
