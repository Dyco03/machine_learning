```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
```

# Dataset

```python
n_features = 12
x,y = make_regression(n_samples= 100, n_features = n_features, noise = 10)
```

```python
print(x.shape)
y = y.reshape(y.shape[0],1)
print(y.shape)
```

```python
X = np.hstack([x,np.ones((x.shape[0],1))])
X.shape
```

```python
theta = np.random.randn(n_features + 1,1)
lamda = 1 # for ridge
```

# modele

```python
def model(X ,theta):
    return X.dot(theta)
```

```python
model(X,theta)
```

# fonction cout (mse)

```python
def cost_function(X,y,theta,lamda):
    m = len(y)
    ridge = lamda * np.sum(theta) **2
    return (1/(2*m)) * (np.sum((model(X,theta) - y) **2) + ridge)
```

```python
cost_function(X,y,theta,lamda)
```

# Gradients et Descente de Gradient

```python
def grad(X,y,theta,lamda):
    m = len(y)
    ridge = lamda * theta
    ridge[-1] = 0
    return (1/m) * (X.T.dot(model(X,theta) - y) + ridge)
```

```python
def gradient_descent(X,y,theta, alpha, n_iteration):
    for i in range(0, n_iteration):
        theta = theta - alpha * grad(X,y,theta,lamda)

    return theta
```

# entrainement

```python
theta_final = gradient_descent(X,y,theta, alpha = 0.1, n_iteration = 1000)
theta_final
```

```python
predictions = model(X,theta_final)
cost_function(X,y,theta_final,lamda)
```

```python
from sklearn.metrics import r2_score

score = r2_score(y,predictions)

print(score)
```
